```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the data from CSV
data = pd.read_csv("question1reddit_twitter_counts.csv")

# Calculate total counts
total_reddit = data['redditCount'].sum()
total_twitter = data['twitterCount'].sum()

# Normalize counts
data['redditCount'] = (data['redditCount'] / total_reddit)*1000
data['twitterCount'] = (data['twitterCount'] / total_twitter)*1000

data['redditCount'] = data['redditCount'].astype(int)
data['twitterCount'] = data['twitterCount'].astype(int)

data.to_csv("question1reddit_twitter_counts_normalized.csv", index=False)
```
