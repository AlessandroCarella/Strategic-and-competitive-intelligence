```python
import pandas as pd
```

```python
reddit = pd.read_csv("question1RedditData.csv")

twitter = pd.read_csv("question1TwitterData.csv")
```

```python
redditCompanies = reddit["company"]
redditOccurrences = reddit["count"]
twitterCompanies = twitter["company"]
twitterOccurrences = twitter["count"]
```

```python
reddit = dict(zip(redditCompanies, redditOccurrences))
twitter = dict(zip(twitterCompanies, twitterOccurrences))
```

```python
mergeDict = {}
for company in reddit:
    if company in twitter:
        mergeDict[company] = {"redditCount":reddit[company], "twitterCount":twitter[company]}
```

```python
mergeDict
```

```python
import csv

df = pd.DataFrame.from_dict(mergeDict, orient='index')

# Reset index to make 'company' a column
df.reset_index(inplace=True)
df.rename(columns={'index': 'company'}, inplace=True)

# Save DataFrame to CSV
df.to_csv('reddit_twitter_counts.csv', index=False)
```
