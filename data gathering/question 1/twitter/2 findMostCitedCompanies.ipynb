to make the computation faster i dedied to rebuild the column with the original tweets as a dictionary:

```python
import string

def create_word_index_dict(list_of_strings):
    word_index_dict = {}
    for i, stringelem in enumerate(list_of_strings):
        words = stringelem.split()
        for word in words:
            # Remove punctuation from the word
            word = word.translate(str.maketrans('', '', string.punctuation))
            if word not in word_index_dict:
                word_index_dict[word] = {i: 1}
            else:
                if i not in word_index_dict[word]:
                    word_index_dict[word][i] = 1
                else:
                    word_index_dict[word][i] += 1
    return word_index_dict

# Example usage:
list_of_strings = ["apple, banana apple", "banana orange", "apple orange banana"]
result = create_word_index_dict(list_of_strings)
print(result)
```

using this method i can change the way i iter through the dataset and make it much faster

```python
wordsInTheDataset = create_word_index_dict(list(df["original_text"]))
```
