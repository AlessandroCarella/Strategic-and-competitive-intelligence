```python
def remove_short_strings(lst):
    return [s for s in lst if len(str(s)) > 2]#some of the names are float or other kind of variables so we just cast them as strings

companiesNamesFiltered1 = remove_short_strings(companiesNames)
len (companiesNamesFiltered1)
```

```python
import string

def create_word_index_dict(list_of_strings):
    word_index_dict = {}
    for i, stringelem in enumerate(list_of_strings):
        words = stringelem.split()
        for word in words:
            # Remove punctuation from the word
            word = word.translate(str.maketrans('', '', string.punctuation))
            if word not in word_index_dict:
                word_index_dict[word] = {i: 1}
            else:
                if i not in word_index_dict[word]:
                    word_index_dict[word][i] = 1
                else:
                    word_index_dict[word][i] += 1
    return word_index_dict
```

```python
wordsInTheDataset = create_word_index_dict(list(df["original_text"]))
```

```python
def findSum (indictionary):
    val = 0
    for k, v in indictionary.items():
        val+=v
    return val

recurrences = {}
for companyName in companiesNamesFiltered1:
    if companyName in wordsInTheDataset:
        recurrences[companyName] = findSum (wordsInTheDataset[companyName])
        print (companyName, recurrences[companyName])
```

```python
# Sorting the dictionary by values
sorted_dict = dict(sorted(recurrences.items(), key=lambda item: item[1], reverse=True))

# Displaying the sorted dictionary
print(sorted_dict)
print (len (sorted_dict))
```

the next approach takes the brown corpus and orders the words in it by the occurrences

once i have this ordered dictionary i want to remove the first n (guessing around 100/1000) words that have the biggest number of occurrances

```python
import nltk
from nltk.corpus import brown
from collections import Counter

# Download the Brown Corpus if not already downloaded
nltk.download('brown')

# Get the words from the Brown Corpus
words = brown.words()

# Count the occurrences of each word
word_counts = Counter(words)

# Sort the words by frequency in descending order
sorted_word_counts = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

# Create a dictionary from the sorted list
word_freq_dict = dict(sorted_word_counts)

print (word_freq_dict)
```

one of the first words that dont seem to appear at the beggining of the word_freq_dict but appears often in the sorted dictionary that we want as output is "next", i will look for the index of it in the dictionary and filter by the ones that are before it 

```python
print(list(word_freq_dict.keys()).index("next"))
```

since it's 267 and we want some room we go for 300

```python
def cut_dictionary(dictionary, n):
    keys = list(dictionary.keys())[:n]
    return {key: dictionary[key] for key in keys}

word_freq_dict_cutted_300 = cut_dictionary(word_freq_dict, 300)
len (word_freq_dict_cutted_300)
```

```python
# Filter out non-company names
company_names_filtered_cutted_300 = {name: count for name, count in sorted_dict.items() if name not in list(word_freq_dict_cutted_300.keys())}
print (company_names_filtered_cutted_300)
print (len(company_names_filtered_cutted_300))
```
