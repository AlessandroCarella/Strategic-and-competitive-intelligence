```python
import pandas as pd
import os
from os import path
import json
import seaborn
import matplotlib as plt
import numpy as np
import math
import requests
import pickle
```

to do that i selected 100 words related to coding

```python
codingWords = [
    "Programming", "Code", "Software", "Development", "Algorithm", "Function", "Variable", "Loop", "Conditional",
    "Class", "Object", "Method", "Function", "Syntax", "Compiler", "Interpreter", "Debugging", "Bug", "Error",
    "Exception", "Comment", "Version control", "Git", "Repository", "Branch", "Merge", "Commit", "Pull request",
    "Push", "Framework", "Library", "Module", "Package", "Dependency", "Dependency injection", "Dependency management",
    "API", "SDK", "IDE", "CLI", "GUI", "Frontend", "Backend", "Full-stack", "Database", "SQL", "NoSQL", "ORM",
    "RESTful", "HTTP", "URL", "JSON", "XML", "HTML", "CSS", "JavaScript", "TypeScript", "Python", "Java", "C++",
    "C#", "Ruby", "PHP", "Swift", "Kotlin", "Rust", "Go", "Perl", "Shell scripting", "Regular expression",
    "Data structure", "Algorithm", "Recursion", "Big O notation", "Sorting algorithm", "Searching algorithm",
    "Graph algorithm", "Binary tree", "Linked list", "Stack", "Queue", "Hash table", "Set", "Map", "Array", "Tuple",
    "List", "Dictionary", "Variable scope", "Global variable", "Local variable", "Constant", "Function parameter",
    "Return value", "Lambda function", "Object-oriented programming", "Inheritance", "Encapsulation", "Polymorphism",
    "Abstraction"
]
```

```python
dfPath = path.join(os.getcwd(), "..", "..", "..", "ChatGPT_Tweets_Dataset_Full.csv")
df = pd.read_csv(dfPath)
```

```python
df.shape
```

and filtered the dataset by the rows: subject, objects and topic

```python
# Create lowercase versions of the columns
df['subjects_lower'] = df['subjects'].str.lower()
df['objects_lower'] = df['objects'].str.lower()
df['topic_lower'] = df['topic'].str.lower()
```

```python
import re

# Escape special characters in codingWords and convert to lowercase
escaped_codingWords = [re.escape(word.lower()) for word in codingWords]

# Join escaped codingWords into a regex pattern with the IGNORECASE flag
pattern = "|".join(escaped_codingWords)
pattern = f"(?i){pattern}"  # Adding the IGNORECASE flag
```

where one of the words of the coding words list is contained in one of the 3

```python
# Apply the regex pattern to the lowercase DataFrame columns
filtered_df = df[df['subjects_lower'].str.contains(pattern) |
                 df['objects_lower'].str.contains(pattern) |
                 df['topic_lower'].str.contains(pattern)]
```

```python
# Drop the lowercase columns if not needed
filtered_df = filtered_df.drop(['subjects_lower', 'objects_lower', 'topic_lower'], axis=1)
```
