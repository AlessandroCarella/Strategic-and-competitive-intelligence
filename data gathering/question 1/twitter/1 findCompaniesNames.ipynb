```python
dfPath = path.join(os.getcwd(), "..", "ChatGPT_Tweets_Dataset_Full.csv")
df = pd.read_csv(dfPath)
```

```python
# Extract unique values from each column
unique_values = {}
for column in df.columns:
    unique_values[column] = df[column].unique()
```

```python
del unique_values["Unnamed: 0"]
del unique_values["tweet_id"]
del unique_values["original_text"]
```

```python
newUniqueValues = {}
for column, values in unique_values.items():
    newUniqueValues[column] = []
    for value in values:
        if isinstance(value, str):   
            newUniqueValues[column].extend(list(set(value.split(", "))))
    newUniqueValues[column] = list(set(newUniqueValues[column]))
```

```python
for column, values in newUniqueValues.items():
    with open(path.join(os.getcwd(), f"{column}_unique_values.txt"), "w", encoding="utf-8") as file:
        for value in values:
            file.write(str(value) + "\n")
```

since the number of rows is too big i decided to filter them out using the size of the company

as reported on the keaggle page the size indicates the number of people in the company and 49% of the companies have just 1 to 10 employees, another 22% only has 11 to 50 and the remaining 29% has more
