```python
def read_articles(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
    technology_count = {}
    for line in lines:
        article = json.loads(line)
        technologies = article['technologies']
        for technology in technologies:
            technology_count[technology] = technology_count.get(technology, 0) + 1
    return technology_count
```

```python
technology_count_Devto_2022 = read_articles("urlsDevTo2022.json")
```

```python
with open('technology_count_Devto_2022.json', 'w') as f:
    json.dump(technology_count_Devto_2022, f)    
```
