```python
#Imports
import requests
from bs4 import BeautifulSoup
from collections import Counter
import spacy
import json
```

```python
#getURL
r = requests.get('https://survey.stackoverflow.co/2022/#most-popular-technologies-language')
```

```python
#checkURLresponse
r.status_code
```

```python
#displayURLresponse
r.text
```

```python
#parseHTML
soup = BeautifulSoup(r.text, 'html.parser')
```

```python
#inspected web page with developer tools to find closest object with ID to data I'm interested in
#saving id in variable
id = 'section-most-popular-technologies-programming-scripting-and-markup-languages'

#find and save html section in new variable
technology_section = soup.find(id=id)
```

```python
#find and save narrower html section in new variable
figure = technology_section.find(id ='most-popular-technologies-language')
```

```python
#display variable content
figure
```

```python
table = figure.find(id = 'languageqmfte')
```

```python
table
```

```python
rows = table.find_all('tr')
```

```python
#data_soup.find_all(attrs={"data-foo": "value"})
technology = []
for table_element in rows:
    tech_stack = table_element.find(attrs={"class": "label"}).text
    technology.append(tech_stack)
```

```python
technology
```

```python
#data_soup.find_all(attrs={"data-foo": "value"})
percentage = []
for table_element in rows:
    tech_stack = table_element.find(attrs={"class": "bar"})['data-percentage']
    percentage.append(tech_stack)
```

```python
percentage
```

```python
len(technology)
```

```python
len(percentage)
```

```python
techStackData2022List = list(zip(technology,percentage))
```
