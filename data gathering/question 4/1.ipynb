### after the second filtering we aim to find the most common sentiment and the most common subjects, objects, predicates and topics

to do that we import the keywords related to the subject of the question

```python
filtered_df = pd.read_csv("filtered_df.csv")
```

```python
with open ("keywordsAndSubFiltered.json") as f:
    keywordsAndSubFiltered = (json.load (f))

remoteWorkKeywordsList = []
for k, v in keywordsAndSubFiltered.items():
    remoteWorkKeywordsList.append(k)
    remoteWorkKeywordsList.extend(v)
```

after we turn the tweets into a dictionary of words where the key is the word and the value is a list of indexes of the dataset where the word appears

```python
from collections import defaultdict

# Initialize dictionary to store word indexes
word_index_dict = defaultdict(list)

# Iterate over each row
for index, row in filtered_df.iterrows():
    # Tokenize the text into words
    words = row['original_text'].split()
    # Update dictionary with word indexes
    for word in words:
        word_index_dict[word].append(index)

# Convert defaultdict to regular dictionary
word_index_dict = dict(word_index_dict)
```

```python
list(word_index_dict.keys())[0], word_index_dict[list(word_index_dict.keys())[0]] #just an example of the word and the relative indexes
```

than we filter the indexes by the ones in the remoteWorkKeywordsList list

```python
indexes = []
for word in remoteWorkKeywordsList:
    if word in word_index_dict:
        indexes.extend(word_index_dict[word])

indexes = list(set(indexes))
len(indexes)
```

```python
filtered_filtered_df = filtered_df.iloc[indexes]
filtered_filtered_df.shape
```

```python
filtered_filtered_df.to_csv("second filtering df.csv", index=False)
```

```python
import pandas as pd
from collections import Counter
import re

# Define the columns you want to analyze
columns = ['sentiment', 'subjects', 'objects', 'predicates', 'topic']

columnsAndWords = {}
# Iterate over each column and count the words
for col in columns:
    secondSplit = []
    for value in list(filtered_filtered_df[col]):
        firstSplit = str(value).split(", ")
        for sentence in firstSplit:
            secondSplit.extend (str(sentence).split())
    columnsAndWords[col] = secondSplit
```

```python
def count_word_occurrences(dictionary):
    new_dict = {}
    for dfColumn, word_list in dictionary.items():
        word_counts = {}
        for word in word_list:
            if word in word_counts:
                word_counts[word] += 1
            else:
                word_counts[word] = 1
        # Sort the word counts dictionary by values in descending order
        sorted_word_counts = dict(sorted(word_counts.items(), key=lambda item: item[1], reverse=True))
        new_dict[dfColumn] = sorted_word_counts
    return new_dict

result = count_word_occurrences(columnsAndWords)
```
