```python
import json
```

```python
with open ("twitterResults.json", "r") as f:
    data = json.load(f)
```

```python
data.keys()
```

```python
topics = data["topic"]
sentiment = data["sentiment"]
objects = data["objects"]
subjects = data["subjects"]
predicates = data["predicates"]
names = ["topics", "sentiment", "subjects", "predicates", "objects"]
datas = [topics, sentiment, subjects, predicates, objects]
```

```python
import csv
for data, name in zip(datas, names):
    headers = [name.replace("s", ""), "count"]
    with open(name + 'Question2Twitter.csv', 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerow(headers)    
        for key, value in data.items():
            writer.writerow([key, value])
```
