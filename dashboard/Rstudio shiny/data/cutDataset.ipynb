```python
import pandas as pd

def cutFirstN(file, n):
    # Load the CSV file into a DataFrame
    df = pd.read_csv(file + ".csv")

    # Drop the first 100 rows
    df = df.iloc[:n]
    
    df.to_csv(file + ".csv", index=False)

files = ["question2ObjectsTwitter", "question2PredicatesTwitter", "question2SentimentTwitter", "question2SubjectsTwitter", "question2TopicsTwitter"]
for file in files:
    cutFirstN(file, 20)
```
