### interesting columns list

```python
goodColumns = ["technologies", "job_competencies", "job_profiles"]
goodColumnsWhenCodeRelated = [
                                #"objects", 
                                #"subjects", 
                                "topic"
                              ]
```

we prompted gpt with this question

"i have to filter a dataset of tweets by the ones related to coding, give me a list of 20 words that could be in tweets related to coding and generative ai"

and the following is the list returned which we will use to filter the dataset on the columns (+ some of ours)

```python
words_related_to_coding_and_ai = [
    "Python",
    "JavaScript",
    "C",
    "cpp",
    "c++",
    "C#",
    "Ruby",
    "Swift",
    "Kotlin",
    "R",
    "Go",
    "Scala",
    "PHP",
    "HTML",
    "CSS",
    "SQL",
    "TypeScript",
    "Perl",
    "Assembly",
    "Objective-C",
    "Programming",
    "Code",
    #"AI",
    #"Machine Learning",
    #"Generative",
    #"Deep Learning",
    #"Neural Networks",
    "Algorithm",
    #"Data Science",
    #"TensorFlow",
    #"PyTorch",
    "GitHub",
    "Framework",
    #"Model",
    #"Development",
    #"Automation",
    #"Creative",
    #"Innovation",
    "Backend",
    "Frontend"
]
```

now we filter the dataset by where the good columns are not null

```python
df = df[df[goodColumns].notnull().all(axis=1)]
df.shape
```

```python
df = df.drop(columns=['Unnamed: 0', "original_text"])
df.to_csv("half filtered.csv", index=False)
```

we still have a big dataset so we can keep the filter and add the next one too

```python
tempDf = df
for column in goodColumnsWhenCodeRelated:
    tempDf = tempDf[tempDf[column].str.contains('|'.join(words_related_to_coding_and_ai).lower(), na=False)]

tempDf.shape
```

we can now save the df to file (we went from 3821843 records to 3806)

```python
tempDf.to_csv("final.csv", index=False)
```
